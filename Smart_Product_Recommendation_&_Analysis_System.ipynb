# Smart Product Recommendation & Analysis System

This mini project retrieves product information from a public REST API
using HTTP GET requests.

The JSON response is converted into a Pandas DataFrame and analyzed
using filtering, sorting, grouping and aggregation.

The system also provides product recommendations based on:
- Price
- Rating
- Stock availability

Finally, the processed data is exported to CSV files.

```python
import requests
import pandas as pd

print("Libraries imported successfully!")
```

```python
api_url = "https://dummyjson.com/products"

parameters = {
    "limit": 100
}

print("API URL:", api_url)
print("Parameters:", parameters)
```

```python
try:
    response = requests.get(
        api_url,
        params=parameters,
        timeout=10
    )

    response.raise_for_status()

    print("Status Code:", response.status_code)
    print("API request successful!")

except requests.RequestException as error:
    print("API request failed:", error)
```

```python
api_data = response.json()

print("JSON data received successfully!")
print("Available keys:", api_data.keys())
```

```python
products = pd.DataFrame(api_data["products"])

print("Number of products:", len(products))
print("Number of columns:", len(products.columns))
```

```python
products = products[
    [
        "id",
        "title",
        "brand",
        "category",
        "price",
        "rating",
        "stock"
    ]
]

products.head(10)
```

```python
print("========== PRODUCT DATASET ==========")

print("Total products :", len(products))
print("Average price  :", round(products["price"].mean(), 2))
print("Average rating :", round(products["rating"].mean(), 2))
print("Total stock    :", products["stock"].sum())
```

```python
search_name = input("Enter product name or keyword: ").strip().lower()

search_result = products[
    products["title"]
    .astype(str)
    .str.lower()
    .str.contains(search_name, na=False)
]

if not search_result.empty:
    print("\n")
    print("\n SEARCH RESULTS")
    print(search_result.to_string(index=False))
else:
    print("\nNo matching product found.")
    print("\nAvailable products:")
    print(products["title"].to_string(index=False))
```

```python
#Filter by maximum price
max_price = float(input("Enter maximum price: "))

price_filtered = products[
    products["price"] <= max_price
]

print("\nProducts within your budget:")
print(price_filtered.to_string(index=False))
```

```python
#Filter by rating
minimum_rating = float(input("Enter minimum rating: "))

rating_filtered = products[
    products["rating"] >= minimum_rating
]

print("\nHighly rated products:")
print(rating_filtered.to_string(index=False))
```

```python
#Show products with stock
available_products = products[
    products["stock"] > 0
]

print("Products currently in stock:")
print(available_products.to_string(index=False))
```

```python
#Sort products by price
price_sorted = products.sort_values(
    by="price",
    ascending=False
)

print("Products from highest price to lowest price:")
print(price_sorted.to_string(index=False))
```

```python
#Sort by rating
rating_sorted = products.sort_values(
    by="rating",
    ascending=False
)

print("Top-rated products:")
print(rating_sorted.head(10).to_string(index=False))
```

```python
#Category analysis
category_summary = products.groupby("category").agg(
    number_of_products=("id", "count"),
    average_price=("price", "mean"),
    average_rating=("rating", "mean"),
    total_stock=("stock", "sum")
).reset_index()

category_summary["average_price"] = category_summary["average_price"].round(2)
category_summary["average_rating"] = category_summary["average_rating"].round(2)

print(category_summary)
```

Allowing the user to enter:

Maximum price

Minimum rating

Minimum stock

```python
max_price = float(input("Enter your maximum budget: "))
min_rating = float(input("Enter minimum rating: "))
min_stock = int(input("Enter minimum stock required: "))

recommended_products = products[
    (products["price"] <= max_price) &
    (products["rating"] >= min_rating) &
    (products["stock"] >= min_stock)
]

recommended_products = recommended_products.sort_values(
    by=["rating", "price"],
    ascending=[False, True]
)

print("\n========== RECOMMENDED PRODUCTS ==========")

if recommended_products.empty:
    print("No products match your requirements.")
else:
    print(recommended_products.to_string(index=False))

print("===========================================")
```

```python
#Create a recommendation score
products["recommendation_score"] = (
    products["rating"] * 20
    + products["stock"] * 0.5
    - products["price"] * 0.2
)

products["recommendation_score"] = products[
    "recommendation_score"
].round(2)

top_recommendations = products.sort_values(
    by="recommendation_score",
    ascending=False
)

print("========== TOP RECOMMENDATIONS ==========")
print(
    top_recommendations[
        [
            "title",
            "category",
            "price",
            "rating",
            "stock",
            "recommendation_score"
        ]
    ].head(10).to_string(index=False)
)
```

```python
#Exporting complete dataset

products.to_csv(
    "all_products.csv",
    index=False
)

print("all_products.csv created successfully!")
```

```python
#Export recommended products
recommended_products.to_csv(
    "recommended_products.csv",
    index=False
)

print("recommended_products.csv created successfully!")
```

```python
#Export category summary
category_summary.to_csv(
    "category_summary.csv",
    index=False
)

print("category_summary.csv created successfully!")
```
